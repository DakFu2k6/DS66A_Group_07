## 1. Cấu hình

```python
import os, time, warnings
os.environ.setdefault("LOKY_MAX_CPU_COUNT", "4")
import numpy as np, pandas as pd
import matplotlib.pyplot as plt, matplotlib.ticker as mtick, seaborn as sns
from scipy import stats
from IPython.display import display, Markdown
import sklearn
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import ParameterGrid
from sklearn.metrics import (average_precision_score, roc_auc_score, precision_recall_curve,
                             precision_score, recall_score, f1_score)

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid", context="notebook")
plt.rcParams.update({"figure.dpi": 110, "axes.titleweight": "bold", "axes.spines.top": False, "axes.spines.right": False})
pd.set_option("display.max_columns", 50); pd.set_option("display.float_format", "{:,.4f}".format)

SEED = 42
rng = np.random.default_rng(SEED)
QUICK_RUN = os.environ.get("QUICK_RUN", "0") == "1"     

TRAIN_END, VAL_END = 119, 149         

REVIEW_COST   = 5.0    # $ cho MỖI giao dịch bị cảnh báo (OTP / step-up / review)
EFFECTIVENESS = 1.0    # tỉ lệ gian lận bị chặn thành công khi đã cảnh báo (1.0 = chặn hết; cận trên)
N_BOOT        = 50 if QUICK_RUN else 500

C_LEG, C_FRA = "#4C78A8", "#E4572E"
COLORS = {"Logistic Regression": "#7F7F7F", "SVM (RBF)": "#9467BD", "Decision Tree": "#2CA02C",
          "Baseline: price": "#17BECF", "Baseline: category": "#BCBD22"}
MODELS    = ["Logistic Regression", "SVM (RBF)", "Decision Tree"]
BASELINES = ["Baseline: price", "Baseline: category"]
ALL = MODELS + BASELINES
print("sklearn", sklearn.__version__, "| pandas", pd.__version__, "| QUICK_RUN =", QUICK_RUN)
```

## 2. Đọc dữ liệu → làm sạch → **chia Train/Val/Test**

| Thao tác | Quyết định | Lý do |
|---|---|---|
| Dấu nháy đơn `'C123'` | bỏ | file gốc bọc mọi chuỗi trong `'...'` |
| `zipcodeOri`, `zipMerchant` | **xóa** (sau khi assert là hằng số) | không mang thông tin |
| `step` | **xóa khỏi bảng dữ liệu** *sau khi* dùng nó để xác định ranh giới chia | theo yêu cầu; dữ liệu đã sắp theo thời gian nên chỉ cần giữ **thứ tự dòng** |
| `age = 'U'` | **thay bằng nhóm 7** | `U` ⇔ gender `E` (Enterprise) – xem Mục 3.4 |
| tiền tố `es_` ở `category` | bỏ | cho dễ đọc |

Tất cả các bước trên **không học gì từ nhãn hay từ phân phối dữ liệu**, nên làm trước khi chia là an toàn. Ngay sau đó dữ liệu được cắt theo thời gian, và **Test bị khoá vào `_VAULT`**.

```python
raw = pd.read_csv('bs140513_032310.csv')
print(f"Đã đọc {'bs140513_032310.csv'}: {raw.shape[0]:,} dòng × {raw.shape[1]} cột")
display(raw.head())

df = raw.copy()
for c in df.select_dtypes(exclude="number").columns:
    df[c] = df[c].astype(str).str.strip("'")
assert df.step.is_monotonic_increasing, "Dữ liệu phải sắp theo thời gian"
assert df.zipcodeOri.nunique() == 1 and df.zipMerchant.nunique() == 1, "zip phải là hằng số"
print("Missing:", int(df.isna().sum().sum()), "| Duplicate:", int(df.duplicated().sum()))

N_TR = int((df.step <= TRAIN_END).sum())
N_VA = int(((df.step > TRAIN_END) & (df.step <= VAL_END)).sum())

df = df.drop(columns=["step", "zipcodeOri", "zipMerchant"])
df["category"] = df.category.str.replace("es_", "", regex=False)
AGE_LABELS = {0: "<=18", 1: "19-25", 2: "26-35", 3: "36-45", 4: "46-55", 5: "56-65", 6: ">65", 7: "Unknown"}
df["age"] = df.age.replace({"U": "7"}).astype(int)         
df = df.reset_index(drop=True)

df_train = df.iloc[:N_TR].copy()
df_val   = df.iloc[N_TR:N_TR + N_VA].copy()
_VAULT   = {"test": df.iloc[N_TR + N_VA:].copy(), "opened": 0}          
del df, raw                                                             

def open_test():
    assert _VAULT["opened"] == 0, "❌ Test đã được mở rồi — không được chạy lần 2!"
    _VAULT["opened"] += 1
    return _VAULT["test"]

split = pd.DataFrame({n: {"steps": s, "giao dịch": len(d), "gian lận": int(d.fraud.sum()),
                          "tỉ lệ gian lận %": 100 * d.fraud.mean(), "$ gian lận": d.loc[d.fraud == 1, "amount"].sum()}
                      for n, d, s in [("Train", df_train, f"0–{TRAIN_END}"), ("Validation", df_val, f"{TRAIN_END+1}–{VAL_END}")]}).T
split.loc["Test 🔒"] = [f"{VAL_END+1}–179", len(_VAULT["test"]), "🔒", "🔒", "🔒"]       
display(split)
assert len(df_train) == N_TR and len(df_val) == N_VA and len(_VAULT["test"]) > 0
```

> **🧠 Vì sao chia theo thời gian, không `train_test_split` ngẫu nhiên?** Mô hình thật luôn dự đoán *tương lai* từ *quá khứ*. Chia ngẫu nhiên sẽ trộn giao dịch của cùng một thẻ bị đánh cắp vào cả Train lẫn Test → mô hình "nhớ" chiến dịch gian lận → điểm đẹp giả tạo. Khách hàng **có** xuất hiện ở cả 3 tập (giống ngân hàng thật: khách cũ vẫn giao dịch), nhưng thời gian không bị đảo.
>
> **Vì sao bỏ `step` mà vẫn chia theo thời gian được?** Ranh giới được tính từ `step` rồi chuyển thành *số dòng*; do file đã sắp theo thời gian (có `assert`), thứ tự dòng chính là thứ tự thời gian. `step` không bao giờ trở thành feature.

### 3.2 Category

```python
cat_tbl = eda.groupby("category").agg(n=("fraud", "size"), frauds=("fraud", "sum"), fraud_rate=("fraud", "mean"))
cat_tbl["share_of_txns"]   = cat_tbl.n / cat_tbl.n.sum()
cat_tbl["share_of_frauds"] = cat_tbl.frauds / cat_tbl.frauds.sum()
cat_tbl["median_amt_legit"] = eda[eda.fraud == 0].groupby("category").amount.median()
cat_tbl["median_amt_fraud"] = eda[eda.fraud == 1].groupby("category").amount.median()
cat_tbl["share_fraud_$"]    = eda[eda.fraud == 1].groupby("category").amount.sum() / eda[eda.fraud == 1].amount.sum()
cat_tbl = cat_tbl.sort_values("fraud_rate", ascending=False)
display(cat_tbl.style.format({"n": "{:,.0f}", "frauds": "{:,.0f}", "fraud_rate": "{:.2%}", "share_of_txns": "{:.2%}", "share_of_frauds": "{:.2%}",
                              "median_amt_legit": "${:,.2f}", "median_amt_fraud": "${:,.2f}", "share_fraud_$": "{:.1%}"}, na_rep="–")
        .bar(subset=["fraud_rate"], color="#F75644"))

ZERO_CATS = list(cat_tbl.index[cat_tbl.frauds == 0])          
RISKY_CATS = list(cat_tbl.index[cat_tbl.frauds > 0])
fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))
p = cat_tbl.sort_values("fraud_rate")
ax[0].barh(p.index, p.fraud_rate, color=C_FRA); ax[0].xaxis.set_major_formatter(mtick.PercentFormatter(1, 0)); ax[0].set_title("Tỉ lệ gian lận theo category")
for y, (r, n) in enumerate(zip(p.fraud_rate, p.n)): ax[0].text(r + .005, y, f"n={n:,}", va="center", fontsize=7)
p2 = cat_tbl.sort_values("share_fraud_$")
ax[1].barh(p2.index, p2["share_fraud_$"].fillna(0), color="darkorange"); ax[1].xaxis.set_major_formatter(mtick.PercentFormatter(1, 0)); ax[1].set_title("% tổng $ gian lận theo category")
plt.tight_layout(); plt.show()
display(Markdown(f"**Category không có gian lận trong Train:** {ZERO_CATS} — chứa **{cat_tbl.loc[ZERO_CATS,'share_of_txns'].sum():.1%}** giao dịch "
                 f"(toàn 'âm tính dễ'). Do đó ROC-AUC bị thổi phồng; PR-AUC và $ mới là thước đo trung thực."))
```

Hai hệ quả: (1) một luật theo category đã là baseline rất mạnh → mô hình ML phải thắng được nó *bên trong* các category rủi ro mới đáng triển khai; (2) tỉ lệ gian lận của category nhỏ được ước lượng từ ít dòng (xem cột `n`) nên kém ổn định. Đây là đặc điểm của **trình mô phỏng**; dữ liệu thật hiếm khi sạch như vậy.

### 3.3 Số tiền giao dịch (`amount`)

```python
overall_auc = roc_auc_score(eda.fraud, eda.amount)
u = stats.mannwhitneyu(eda.loc[eda.fraud == 1, "amount"], eda.loc[eda.fraud == 0, "amount"])
print(f"Chỉ riêng amount: ROC-AUC = {overall_auc:.3f} (Mann-Whitney p = {u.pvalue:.1e})")
print(eda.groupby("fraud").amount.describe().rename(index={0: "hợp lệ", 1: "gian lận"}).round(2))

rows = []
for c in [c for c in RISKY_CATS if cat_tbl.loc[c, "frauds"] >= 50 and cat_tbl.loc[c, "frauds"] < cat_tbl.loc[c, "n"]]:
    g = eda[eda.category == c]
    rows.append({"category": c, "median_legit": g[g.fraud == 0].amount.median(), "median_fraud": g[g.fraud == 1].amount.median(),
                 "AUC_amount_within_category": roc_auc_score(g.fraud, g.amount)})
within = pd.DataFrame(rows).set_index("category"); within["tỉ số median fraud/legit"] = within.median_fraud / within.median_legit
print("\nAmount có còn phân biệt gian lận BÊN TRONG từng category không? (kiểm soát confounding theo category)")
display(within.sort_values("AUC_amount_within_category", ascending=False))

fig, ax = plt.subplots(1, 2, figsize=(14, 4.2))
d2 = eda.assign(log10_amount=np.log10(eda.amount.clip(lower=0.5)), nhan=eda.fraud.map({0: "Hợp lệ", 1: "Gian lận"}))
sns.histplot(data=d2, x="log10_amount", hue="nhan", stat="density", common_norm=False, element="step", bins=60,
             palette={"Hợp lệ": C_LEG, "Gian lận": C_FRA}, ax=ax[0]); ax[0].set_title("Phân phối log10(amount) theo nhãn")
sns.boxplot(data=d2[d2.category.isin(within.index)], x="category", y="amount", hue="nhan", showfliers=False,
            palette={"Hợp lệ": C_LEG, "Gian lận": C_FRA}, ax=ax[1])
ax[1].set_yscale("log"); ax[1].tick_params(axis="x", rotation=60); ax[1].set_xlabel(""); ax[1].set_title("Amount theo category & nhãn")
plt.tight_layout(); plt.show()
```

**🧠 Đọc kết quả.** Nếu AUC của `amount` *trong từng category* vẫn cao, amount là tín hiệu **độc lập** với category (không chỉ là "travel vốn đắt"). Cơ chế trong mô phỏng là "kẻ gian mua đồ đắt"; ngoài đời kẻ gian thường *thử thẻ bằng khoản nhỏ* — một mô hình học "to = gian lận" sẽ bỏ sót kiểu đó. Cũng vì thế baseline **Price** (chỉ dùng amount) là đối thủ đáng gờm.

### 3.4 Nhân khẩu học (age, gender) và kiểm tra confounding

```python
def cramers_v(x, y):
    ct = pd.crosstab(x, y); chi2, p, _, _ = stats.chi2_contingency(ct)
    return float(np.sqrt(chi2 / (ct.values.sum() * (min(ct.shape) - 1)))), float(p)
rows = []
for c in ["category", "merchant", "age", "gender"]:
    v, p = cramers_v(eda[c], eda.fraud); rows.append({"feature": c, "số mức": eda[c].nunique(), "Cramers_V": v, "chi2_p": p})
display(pd.DataFrame(rows).set_index("feature").style.format({"Cramers_V": "{:.3f}", "chi2_p": "{:.1e}"}))

chk = pd.crosstab(eda.age.eq(7), eda.gender.eq("E"))
print("age==7 (Unknown) ⇔ gender=='E' (Enterprise)?  →", bool((eda.age.eq(7) == eda.gender.eq("E")).all()))

zc = eda.category.isin(ZERO_CATS)
def demo_table(col):
    g = eda.groupby(col)
    t = pd.DataFrame({"n": g.size(), "tỉ lệ gian lận thô": g.fraud.mean(), "% nằm trong cat sạch": zc.groupby(eda[col]).mean()})
    t["tỉ lệ GL trong cat rủi ro"] = eda[~zc].groupby(col).fraud.mean(); return t
gender_t, age_t = demo_table("gender"), demo_table("age").rename(index=AGE_LABELS)
fmt = {"n": "{:,.0f}", "tỉ lệ gian lận thô": "{:.2%}", "% nằm trong cat sạch": "{:.1%}", "tỉ lệ GL trong cat rủi ro": "{:.2%}"}
print("Gender"); display(gender_t.style.format(fmt, na_rep="–")); print("Age"); display(age_t.style.format(fmt, na_rep="–"))
fig, ax = plt.subplots(1, 2, figsize=(12, 3.6))
for a, t, title in [(ax[0], gender_t, "Gender"), (ax[1], age_t, "Age group")]:
    x = np.arange(len(t)); w = .4
    a.bar(x - w/2, t["tỉ lệ gian lận thô"], w, label="thô", color="#999"); a.bar(x + w/2, t["tỉ lệ GL trong cat rủi ro"].fillna(0), w, label="chỉ cat rủi ro", color=C_FRA)
    a.set_xticks(x); a.set_xticklabels(t.index); a.yaxis.set_major_formatter(mtick.PercentFormatter(1, 1)); a.set_title(f"Tỉ lệ gian lận theo {title}")
ax[0].legend(); plt.tight_layout(); plt.show()
```

**🧠 Đọc kết quả.** Với hàng trăm nghìn dòng, *ý nghĩa thống kê ≠ tầm quan trọng*: nhìn **Cramér's V** (độ lớn hiệu ứng) chứ đừng chỉ nhìn p-value. Cột "% nằm trong cat sạch" cho biết chênh lệch giữa các nhóm có phải do **cơ cấu category** (confounding) hay không. Nhóm `age = 7` trùng hoàn toàn với gender `E`, nên việc thay `U → 7` chỉ là *đặt tên*, không phải thông tin mới; vì 7 không phải thứ tự tuổi thật nên age được **one-hot** (không đưa như số liên tục) cho mô hình tuyến tính/SVM. Dù thấy chênh lệch theo giới, mô hình **không nên** biến nó thành luật "ai rủi ro hơn" (rủi ro công bằng/đạo đức) — Mục 6.3 đo xem bỏ age/gender có làm tệ đi không.

### 3.5 Merchant

```python
mer = eda.groupby("merchant").agg(category=("category", "first"), n=("fraud", "size"), frauds=("fraud", "sum"), rate=("fraud", "mean"))
mer["category_rate"] = mer.category.map(cat_tbl.fraud_rate); mer["vượt so với category"] = mer.rate - mer.category_rate
print(f"Merchant có ≥1 gian lận: {(mer.frauds>0).sum()}/{len(mer)} | top-5 merchant chứa {100*mer.frauds.nlargest(5).sum()/mer.frauds.sum():.0f}% số vụ")
print("Mỗi merchant thuộc đúng 1 category:", bool((eda.groupby('merchant').category.nunique() == 1).all()))
display(mer[mer.n >= 100].sort_values("rate", ascending=False).head(8).style.format({"n": "{:,.0f}", "rate": "{:.1%}", "category_rate": "{:.1%}", "vượt so với category": "{:+.1%}"}))

rows = []
for c, g in eda[~eda.category.isin(ZERO_CATS)].groupby("category"):
    ct = pd.crosstab(g.merchant, g.fraud)
    if ct.shape[0] > 1 and ct.shape[1] == 2:
        rows.append({"category": c, "số merchant": ct.shape[0], "chi2_p (merchant × fraud)": stats.chi2_contingency(ct)[1]})
print("\nTrong cùng 1 category, merchant có còn quan trọng?")
display(pd.DataFrame(rows).set_index("category").style.format({"chi2_p (merchant × fraud)": "{:.1e}"}))
fig, ax = plt.subplots(figsize=(10, 3.8))
sns.stripplot(data=mer[~mer.category.isin(ZERO_CATS)], x="category", y="rate", size=7, color=C_FRA, ax=ax)
ax.yaxis.set_major_formatter(mtick.PercentFormatter(1, 0)); ax.tick_params(axis="x", rotation=60); ax.set_xlabel(""); ax.set_title("Tỉ lệ gian lận theo merchant trong từng category rủi ro")
plt.tight_layout(); plt.show()
```

**🧠 Đọc kết quả.** Nếu merchant còn giải thích gian lận *bên trong* category thì nó là tín hiệu mạnh — nhưng cũng là **rủi ro triển khai**: mô hình "thuộc lòng" merchant ID không chấm được merchant mới và dễ bị kẻ gian né bằng cách đổi merchant. Mục 6.3 (ablation) định lượng mức phụ thuộc này. Không dùng *target encoding* cho merchant — dùng one-hot để tránh rò rỉ nhãn qua mã hoá.

## 4. Feature engineering (chỉ dùng quá khứ) + 🔍 **kiểm toán rò rỉ**

| Feature | Ý nghĩa | Fit-trên-Train? |
|---|---|---|
| `log_amount` | log(1+amount) | – |
| `amt_z_in_category` | amount bất thường thế nào so với category (z-score của log-amount) | ✅ mean/std **fit trên Train** |
| `log_n_prior_txn`, `is_first_txn` | khách có bao nhiêu lịch sử | – (chỉ quá khứ) |
| `log_amt_to_cust_mean`, `log_amt_to_cust_max` | amount so với trung bình/max **quá khứ** của khách | – |
| `is_new_merchant`, `is_new_category` | lần đầu khách này mua ở merchant/category (khi đã có lịch sử) | – |
| `category`, `merchant`, `gender`, `age` | one-hot | ✅ encoder fit trên Train |

*Cố ý không có:* `customer` ID, `step`, cờ "từng gian lận". Feature theo `step` (cùng ngày, số ngày từ lần trước) cũng không dùng vì `step` đã bị xóa.

```python
def add_history_features(d):
    d = d.copy(); g = d.groupby("customer", sort=False)
    d["n_prior_txn"] = g.cumcount()
    d["is_first_txn"] = (d.n_prior_txn == 0).astype(int)
    prior_mean = (g.amount.cumsum() - d.amount) / d.n_prior_txn.replace(0, np.nan)
    d["log_amt_to_cust_mean"] = np.log((d.amount + 1) / (prior_mean + 1)).fillna(0.0)
    prior_max = g.amount.cummax().groupby(d.customer, sort=False).shift(1)
    d["log_amt_to_cust_max"] = np.log((d.amount + 1) / (prior_max + 1)).fillna(0.0)
    d["is_new_merchant"] = ((d.groupby(["customer", "merchant"], sort=False).cumcount() == 0) & (d.n_prior_txn > 0)).astype(int)
    d["is_new_category"] = ((d.groupby(["customer", "category"], sort=False).cumcount() == 0) & (d.n_prior_txn > 0)).astype(int)
    d["log_n_prior_txn"] = np.log1p(d.n_prior_txn)
    d["log_amount"] = np.log1p(d.amount)
    return d

BEH_COLS = ["log_n_prior_txn", "is_first_txn", "log_amt_to_cust_mean", "log_amt_to_cust_max", "is_new_merchant", "is_new_category"]

_tr_log = np.log1p(df_train.amount)
CAT_STATS = _tr_log.groupby(df_train.category).agg(["mean", "std"]).fillna({"std": 1.0})
def add_amt_z(d):
    d = d.copy()
    d["amt_z_in_category"] = ((d.log_amount - d.category.map(CAT_STATS["mean"])) / d.category.map(CAT_STATS["std"]).clip(lower=1e-6)).fillna(0.0)
    return d

tv = add_history_features(pd.concat([df_train, df_val], ignore_index=True))
tr = add_amt_z(tv.iloc[:len(df_train)]).reset_index(drop=True)
va = add_amt_z(tv.iloc[len(df_train):]).reset_index(drop=True)
print("Train:", tr.shape, "| Val:", va.shape)

src = pd.concat([df_train, df_val], ignore_index=True)
for cut_at in [len(df_train) // 2, len(df_train), len(df_train) + len(df_val) // 2]:
    cut = add_history_features(src.iloc[:cut_at])
    diff = (cut[BEH_COLS + ["log_amount"]].values - tv.iloc[:cut_at][BEH_COLS + ["log_amount"]].values)
    assert np.abs(diff).max() == 0, f"RÒ RỈ tại cut={cut_at}!"
assert len(tr) == N_TR and len(va) == N_VA
print("✅ Kiểm toán: feature 'chỉ-thấy-quá-khứ' ≡ feature 'thấy-cả-tương-lai' tại 3 điểm cắt → không rò rỉ.")
display(tr.groupby("fraud")[["log_amount", "amt_z_in_category"] + BEH_COLS].mean().rename(index={0: "hợp lệ", 1: "gian lận"}).T)
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 3.6))
sns.kdeplot(data=tr.sample(min(150_000, len(tr)), random_state=SEED), x="log_amt_to_cust_mean", hue="fraud", common_norm=False,
            palette={0: C_LEG, 1: C_FRA}, ax=ax[0], bw_adjust=1.5, clip=(-4, 6)); ax[0].set_title("Amount so với trung bình quá khứ của khách")
nm = tr.groupby("fraud")[["is_new_merchant", "is_new_category", "is_first_txn"]].mean().rename(index={0: "hợp lệ", 1: "gian lận"})
nm.plot.bar(ax=ax[1], rot=0, color=["#4C78A8", "#72B7B2", "#BBBBBB"]); ax[1].set_title("Cờ 'mới lạ' (tỉ lệ)")
sns.kdeplot(data=tr.sample(min(150_000, len(tr)), random_state=SEED), x="amt_z_in_category", hue="fraud", common_norm=False,
            palette={0: C_LEG, 1: C_FRA}, ax=ax[2], bw_adjust=1.5, clip=(-4, 6)); ax[2].set_title("Độ bất thường của amount trong category")
plt.tight_layout(); plt.show()
```

## 5. Khung đánh giá & hàm tiền

**Vì sao không accuracy / ROC-AUC?** Gian lận ~1.2% và ~90% dòng nằm trong category sạch ⇒ ROC-AUC luôn rất cao. **PR-AUC (Average Precision)** tập trung vào lớp gian lận — đây là metric không-phụ-thuộc-ngưỡng chính.

**Metric theo ngưỡng (Precision / Recall / F1)** dùng ngưỡng **tối đa hoá F1 trên Validation**.

**Tiền tiết kiệm:**
$$\text{cost}=\underbrace{\sum_{\text{gian lận lọt}}\text{amount}}_{\text{tổn thất}}+c_{\text{review}}\times\#\text{cảnh báo}\qquad\text{Tiết kiệm}=\text{tổn thất nếu không có mô hình}-\text{cost có mô hình}$$
với $c_{review}=\$5$/cảnh báo (**giả định**, có phân tích độ nhạy). Ngưỡng **tối thiểu hoá cost trên Validation** là ngưỡng triển khai.

**Hai baseline** mà mô hình ML phải đánh bại: ① *Price*: điểm = `amount` · ② *Category*: điểm = tỉ lệ gian lận của category trong Train.

```python
def cost_opt_threshold(y, s, amt):
    o = np.argsort(-s, kind="stable"); ss, ys, am = s[o], y[o], amt[o]
    total = float((y * amt).sum())
    cost = total - EFFECTIVENESS * np.cumsum(ys * am) + REVIEW_COST * np.arange(1, len(s) + 1)
    last = np.flatnonzero(np.r_[ss[1:] != ss[:-1], True])         
    j = last[np.argmin(cost[last])]
    return np.inf if cost[j] >= total else ss[j]

def f1_opt_threshold(y, s):
    P, R, T = precision_recall_curve(y, s)
    f1 = 2 * P[:-1] * R[:-1] / np.clip(P[:-1] + R[:-1], 1e-12, None)
    return T[np.argmax(f1)]

def money(y, amt, flag):
    flag = np.asarray(flag, bool); loss = float(amt[y == 1].sum())
    prevented = EFFECTIVENESS * float(amt[flag & (y == 1)].sum()); review = REVIEW_COST * int(flag.sum())
    return dict(loss_no_model=loss, prevented=prevented, review_cost=review, net_saved=prevented - review, saved_pct=(prevented - review) / loss)

def evaluate(y, s, amt, f1_thr, cost_thr):
    pf = s >= f1_thr; pc = s >= cost_thr
    return {"PR_AUC": average_precision_score(y, s), "ROC_AUC": roc_auc_score(y, s),
            "precision@F1thr": precision_score(y, pf, zero_division=0), "recall@F1thr": recall_score(y, pf), "F1": f1_score(y, pf, zero_division=0),
            "alerts@costthr": int(pc.sum()), "recall@costthr": recall_score(y, pc), **money(y, amt, pc)}

yva, ytr = va.fraud.values, tr.fraud.values
amt_va = va.amount.values
```

### 5.1 Ma trận đặc trưng (fit trên Train) & huấn luyện
Mọi mô hình dùng chung một pipeline tiền xử lý: `StandardScaler` cho số + `OneHotEncoder` cho category/merchant/gender/age (age là **categorical**, nhóm 7 = Unknown). Cả hai **fit trên Train**.

**SVM (RBF)** có độ phức tạp ~O(n²) → không chạy nổi trên ~400k dòng. SVM được huấn luyện trên **tập con under-sample của Train** (toàn bộ gian lận Train + 40,000 giao dịch hợp lệ ngẫu nhiên *lấy từ Train*), rồi chấm điểm toàn bộ Val/Test bằng `decision_function`. Vì ngưỡng được **hiệu chỉnh lại trên Val**, việc under-sample không làm sai quyết định cuối; đánh đổi là SVM thấy ít dữ liệu hợp lệ hơn.

```python
CATF = ["category", "merchant", "gender", "age"]
NUM_STATIC = ["log_amount", "amt_z_in_category"]
NUM_ALL = NUM_STATIC + BEH_COLS
for d in (tr, va): d["age"] = d["age"].astype(str)         

def make_data(num, cat):
    pre = ColumnTransformer([("num", StandardScaler(), num),
                             ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False, dtype=np.float32), cat)]) if cat else \
          ColumnTransformer([("num", StandardScaler(), num)])
    Xtr = pre.fit(tr[num + cat]).transform(tr[num + cat]).astype(np.float32)        # FIT CHỈ TRÊN TRAIN
    Xva = pre.transform(va[num + cat]).astype(np.float32)
    return pre, Xtr, Xva

n_leg = 4_000 if QUICK_RUN else 40_000
SVM_IDX = np.sort(np.concatenate([np.flatnonzero(ytr == 1), rng.choice(np.flatnonzero(ytr == 0), size=n_leg, replace=False)]))
print(f"SVM train subset: {len(SVM_IDX):,} dòng ({ytr[SVM_IDX].mean():.1%} gian lận) — lấy từ Train")

FACTORY = {
    "Logistic Regression": lambda **p: LogisticRegression(max_iter=500, random_state=SEED, **p),
    "SVM (RBF)":           lambda **p: SVC(kernel="rbf", cache_size=1000, random_state=SEED, **p),
    "Decision Tree":       lambda **p: DecisionTreeClassifier(random_state=SEED, **p),
}
def score(m, X): return m.predict_proba(X)[:, 1] if hasattr(m, "predict_proba") else m.decision_function(X)
def fit_score(name, params, Xtr_, Xva_):
    idx = SVM_IDX if name.startswith("SVM") else slice(None)
    m = FACTORY[name](**params).fit(Xtr_[idx], ytr[idx]); return m, score(m, Xva_)
```

### 5.2 Thí nghiệm 1 — feature hành vi có giúp không? 

```python
DEFAULTS = {"Logistic Regression": dict(C=1.0), "SVM (RBF)": dict(C=1.0, gamma="scale"), "Decision Tree": dict(max_depth=8, min_samples_leaf=50)}
FEATURE_SETS = {"static": NUM_STATIC, "static + behavioural": NUM_ALL}
rows = []
for fs, num in FEATURE_SETS.items():
    _, Xa, Xb = make_data(num, CATF)
    for n in MODELS:
        t = time.time(); _, s = fit_score(n, DEFAULTS[n], Xa, Xb)
        rows.append({"feature_set": fs, "model": n, "val_PR_AUC": average_precision_score(yva, s), "fit+score (s)": time.time() - t})
exp1 = pd.DataFrame(rows)
piv = exp1.pivot(index="model", columns="feature_set", values="val_PR_AUC")[list(FEATURE_SETS)]; piv["gain"] = piv["static + behavioural"] - piv["static"]
display(piv.style.format("{:.4f}").background_gradient(cmap="Greens", subset=["gain"]))
FINAL_NUM = FEATURE_SETS[exp1.groupby("feature_set").val_PR_AUC.mean().idxmax()]
print("Bộ feature đưa vào bước sau:", "static + behavioural" if FINAL_NUM is NUM_ALL else "static")
ax = piv[list(FEATURE_SETS)].plot.bar(figsize=(8, 3.5), color=["#BBB", C_FRA], rot=0); ax.set_ylabel("PR-AUC (Val)"); ax.set_title("Tác động của feature hành vi"); plt.tight_layout(); plt.show()
```

> **🧠** Thêm feature chưa chắc đã tốt hơn — ta *đo* thay vì giả định. Nếu mọi mô hình đều cải thiện, tín hiệu mới là thật chứ không phải đặc thù của một thuật toán. Quyết định chọn bộ feature được đưa ra **trên Val**.

### 5.3 Tinh chỉnh tham số (trên Validation) và vai trò của `class_weight`
Lưới nhỏ có chủ đích — mục tiêu là mô hình hợp lý, được regularise, không phải "vắt kiệt leaderboard". `class_weight` nằm trong lưới nên câu hỏi mất cân bằng được trả lời bằng thực nghiệm. **Không dùng SMOTE**: nội suy giữa các one-hot/tỉ lệ hành vi tạo giao dịch giả phi thực tế. Không dùng k-fold ngẫu nhiên (sẽ rò rỉ tương lai) — dùng đúng cửa sổ Val theo thời gian.

```python
pre, Xtr, Xva = make_data(FINAL_NUM, CATF)
feat_names = pre.get_feature_names_out()
print("Ma trận đặc trưng:", Xtr.shape, "→ Val:", Xva.shape)

GRIDS = {
    "Logistic Regression": {"C": [0.1, 1.0, 10.0], "class_weight": [None, "balanced"]},
    "SVM (RBF)":           {"C": [1.0, 10.0], "gamma": ["scale", 0.02]},
    "Decision Tree":       {"max_depth": [4, 6, 8, 12], "min_samples_leaf": [20, 100], "class_weight": [None, "balanced"]},
}
if QUICK_RUN: GRIDS = {k: dict(list(v.items())[:1]) for k, v in GRIDS.items()}

tuning, best_cfg, models_, val_scores = {}, {}, {}, {}
for n, grid in GRIDS.items():
    t0 = time.time(); rows, best = [], (-1, None, None, None)
    for p in ParameterGrid(grid):
        m, s = fit_score(n, p, Xtr, Xva); ap = average_precision_score(yva, s)
        rows.append({**{k: str(v) for k, v in p.items()}, "val_PR_AUC": ap})
        if ap > best[0]: best = (ap, p, m, s)
    tuning[n] = pd.DataFrame(rows).sort_values("val_PR_AUC", ascending=False)
    val_scores[n], best_cfg[n], models_[n] = best[3], best[1], best[2]
    print(f"{n:20s} {len(rows):2d} cấu hình, {time.time()-t0:5.0f}s → best val PR-AUC = {best[0]:.4f}  {best[1]}")
for n, t in tuning.items():
    print(f"\n── {n}: top cấu hình"); display(t.head(4))
```

> **🧠** Lưới nhỏ & chỉ một cửa sổ Val ⇒ chênh nhau vài phần nghìn giữa các cấu hình là *nhiễu*, đừng diễn giải quá mức. Xem bảng trên để biết `class_weight='balanced'` có thực sự giúp PR-AUC hay chỉ làm méo xác suất.

### 5.4 Điểm của baseline & ngưỡng (chọn trên Validation → đóng băng)

```python
sign = 1.0 if roc_auc_score(ytr, tr.amount) >= 0.5 else -1.0
print(f"Baseline Price: AUC(amount) trên Train = {roc_auc_score(ytr, tr.amount):.3f} → chiều điểm = {'+' if sign > 0 else '-'}amount")

cat_rate = tr.groupby("category").fraud.mean()
val_scores["Baseline: price"]    = sign * va.amount.values
val_scores["Baseline: category"] = va.category.map(cat_rate).fillna(0).values

F1_THR   = {n: f1_opt_threshold(yva, val_scores[n]) for n in ALL}
COST_THR = {n: cost_opt_threshold(yva, val_scores[n], amt_va) for n in ALL}
val_res = pd.DataFrame({n: evaluate(yva, val_scores[n], amt_va, F1_THR[n], COST_THR[n]) for n in ALL}).T
print(f"VALIDATION — tổn thất nếu KHÔNG có mô hình = ${val_res.loss_no_model.iloc[0]:,.0f}")
display(val_res[["PR_AUC", "F1", "precision@F1thr", "recall@F1thr", "alerts@costthr", "net_saved", "saved_pct"]]
        .style.format({"PR_AUC": "{:.4f}", "F1": "{:.3f}", "precision@F1thr": "{:.3f}", "recall@F1thr": "{:.3f}", "alerts@costthr": "{:,.0f}", "net_saved": "${:,.0f}", "saved_pct": "{:.1%}"})
        .highlight_max(subset=["PR_AUC", "F1", "net_saved"], color="#CDEBC5"))


BEST = val_res.loc[MODELS, "PR_AUC"].astype(float).idxmax()
BEST_BASE = val_res.loc[BASELINES, "net_saved"].astype(float).idxmax()
print(f"\n🏆 Mô hình chọn theo Validation PR-AUC: {BEST}   |   baseline mạnh nhất theo Val ($): {BEST_BASE}")
```

> **🧠 Cách đọc bảng Validation.**

### 8.2 Độ bất định: bootstrap theo khối thời gian
Giao dịch cùng ngày tương quan nhau (cùng chiến dịch gian lận), nên ta resample **khối giao dịch liên tiếp** (Test được chia thành 30 khối liên tiếp ≈ 30 ngày; vì `step` đã bị xóa, khối theo thứ tự dòng là proxy của "ngày") thay vì từng dòng. Ngưỡng vẫn là ngưỡng đã đóng băng từ Val. Cho khoảng tin cậy 95% và **hiệu paired** giữa các mô hình/baseline.

```python
blocks = np.array_split(np.arange(len(te)), 30)
rng_b = np.random.default_rng(SEED)
B = {k: {n: [] for n in ALL} for k in ["PR_AUC", "F1", "net_saved"]}
for b in range(N_BOOT):
    ix = np.concatenate([blocks[i] for i in rng_b.choice(len(blocks), len(blocks), replace=True)])
    yb, ab = yte[ix], amt_te[ix]
    for n in ALL:
        s = test_scores[n][ix]
        B["PR_AUC"][n].append(average_precision_score(yb, s))
        B["F1"][n].append(f1_score(yb, s >= F1_THR[n], zero_division=0))
        B["net_saved"][n].append(money(yb, ab, s >= COST_THR[n])["net_saved"])
B = {k: {n: np.array(v) for n, v in d.items()} for k, d in B.items()}
ci = pd.DataFrame({(k, q): {n: np.percentile(B[k][n], p) for n in ALL} for k in B for q, p in [("lo", 2.5), ("hi", 97.5)]})
point = test_res[["PR_AUC", "F1", "net_saved"]].astype(float)
summ = pd.DataFrame({f"{k} [95% CI]": [f"{point.loc[n,k]:,.4f} [{ci[(k,'lo')][n]:,.4f}, {ci[(k,'hi')][n]:,.4f}]" if k != "net_saved"
                                      else f"${point.loc[n,k]:,.0f} [${ci[(k,'lo')][n]:,.0f}, ${ci[(k,'hi')][n]:,.0f}]" for n in ALL] for k in B}, index=ALL)
display(summ)

def paired(a, b, k):
    d = B[k][a] - B[k][b]; return {"mean_diff": d.mean(), "CI_lo": np.percentile(d, 2.5), "CI_hi": np.percentile(d, 97.5), "P(a>b)": (d > 0).mean()}
TOP_BASE = max(BASELINES, key=lambda n: point.loc[n, "net_saved"])
rows = []
for k in ["PR_AUC", "F1", "net_saved"]:
    for a, b in [(BEST, TOP_BASE)] + [(BEST, m) for m in MODELS if m != BEST]:
        rows.append({"metric": k, "so sánh": f"{a}  −  {b}", **paired(a, b, k)})
print("Hiệu paired (dương = vế trái tốt hơn):"); display(pd.DataFrame(rows).set_index(["metric", "so sánh"]).style.format("{:,.4f}"))
```

6. **Hướng cải tiến (không làm ở đây để tránh chọn lại trên Test):** refit trên Train+Val với siêu tham số cố định, gradient boosting, feature cửa sổ thời gian, hiệu chuẩn xác suất, ngưỡng theo năng lực đội review.
